```python
g = sns.FacetGrid(df, col="Region", hue="Producto", height=4, aspect=0.8)
g.map(sns.scatterplot, "Mes", "Ventas")
g.set_axis_labels("Mes", "Ventas")
g.set_titles(col_template="Región {col_name}")
g.add_legend(title="Producto", bbox_to_anchor=(1, 0.5), loc="center left")# parámaetro bbox_to_anchor para mover la leyenda de los colores dentro del canvas pero fuera del grafico
# para rotar los ticks hacemos un recorrido por los 4 graficos y le pasamos la funcion tick_params() y de parametros el eje a rotar y el grado de rotación. hay 2 formas
# primero veamos tick_params que modifica los parametros de la figura.
for ax in g.axes.flat:
    ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show();
```

```python
g = sns.FacetGrid(df, col="Region", hue="Producto", height=4, aspect=0.8)
g.map(sns.scatterplot, "Mes", "Ventas")
g.set_axis_labels("Mes", "Ventas")
g.set_titles(col_template="Región {col_name}")
g.add_legend(title="Producto", bbox_to_anchor=(1, 0.5), loc="center left")
# ahora veamos la 2da forma, que es seteando las propiedades de la figura y nos permite alinear las etiquetas tmb con el parametro 'ha' y queda un poco mejor
for ax in g.axes.flat:
    plt.setp(ax.get_xticklabels(), rotation=45, ha="right")

plt.tight_layout()
plt.show();
```

##Segundo  
Graficamos las distribuciones con sns.displot  
Debemos especificar la columna de valores en el parámetro x, en este caso nos interesa la columna pulse
Como queremos separar las distribuciones según el tipo de actividad, pasamos el parámetro hue='kind'

```python
plt.figure()
# Figure-level para eso usamos distplot
ax = sns.displot(data=df_30_min, x='pulse', kind='kde', hue='kind', fill=True)# kind es el tipo de grafico, kde es de densidad (kernel density estimate) hue es para colorearlos segun 'kind' tipo de ejercicio
# fill=true es para que pinte el area debajo de las curvas para que sea más visual
ax.set(xlabel='Frecuencia Cardíaca', ylabel='Densidad', title='Distribución  de las pulsaciones');# grabamos las etiquetas de los ejes x e y. Y ponemos un titulo.
```

```python
plt.figure()
# Figure-level para eso usamos distplot
ax = sns.displot(data=df_30_min, x='pulse', kind='hist', hue='kind', fill=True)
ax.set(xlabel='Frecuencia Cardíaca', ylabel='Densidad', title='Distribución  de las pulsaciones');
```

```python
plt.figure()
# Figure-level para eso usamos distplot
ax = sns.displot(data=df_30_min, x='pulse', kind='kde', hue='diet', fill=True)
ax.set(xlabel='Frecuencia Cardíaca', ylabel='Densidad', title='Distribución  de las pulsaciones');
```

```python
ax = sns.catplot(data=df_ejercicio, kind='violin', x='time', y='pulse', hue='diet', split=True)
ax.set(xlabel='Duración de ejercicio', ylabel='Frecuencia cardíaca', title='Categorización de la distribución de pulsaciones')
```

El equipo directivo está interesado en comprender mejor cómo se relacionan las ventas, la satisfacción del cliente y los costos operativos en estas cuatro regiones. Con estos insights, buscan tomar decisiones informadas sobre dónde invertir más recursos, cuáles productos necesitan mayor atención y cómo mejorar la satisfacción del cliente.

**Consigna:** Diseñar un plan de visualización que permita al equipo directivo analizar y comprender mejor los datos, enfocándose en las siguientes preguntas clave:  
1- Ventas y Regiones: ¿Cómo se comparan las ventas de los diferentes productos en las cuatro regiones a lo largo del tiempo? ¿Hay alguna región que se destaque por un rendimiento superior o inferior?

1- Ventas y Regiones: ¿Cómo se comparan las ventas de los diferentes productos en las cuatro regiones a lo largo del tiempo? ¿Hay alguna región que se destaque por un rendimiento superior o inferior?

```python
g = sns.FacetGrid(df, col="Region", hue="Producto", height=4, aspect=1)
g.map(sns.scatterplot, "Mes", "Ventas")
g.set_titles(col_template="Región {col_name}")
g.add_legend()
plt.show();
```

```python
g = sns.FacetGrid(df, col="Region", hue="Producto", height=4, aspect=1)
g.map(sns.lineplot, "Mes", "Ventas")
g.set_titles(col_template="Región {col_name}")
g.add_legend()
plt.show();
```

### respuesta:
Productos:  
Producto A: Region Norte, Mal | Region Sur, Muy bien | Region Este, Muy bien | Region Oeste, Mal  
Producto B: Region Norte, Mal | Region Sur, Masomeno' | Region Este, Mal | Region Oeste, Muy bien  
Producto C: Region Norte, Muy bien | Region Sur, Mal | Region Este, Muy bien | Region Oeste, Muy bien  

Region Norte: 1 - 2  
Region Sur: 1 - 1 - 1  
Region Este: 2 - 1  
Region Oeste: 2 - 1  

Conclusión:  
La regiones Este y Oeste son las que tienen más ventas  
Los productos que mejor desempeño muestran son: C con el mayor desempeño seguido por el producto A que se desmepeño bien en 2 regiones.

2- Satisfacción del Cliente: ¿Existe alguna relación entre las ventas de un producto y la satisfacción del cliente? ¿Cómo varía esta relación entre las diferentes regiones?

```python
g = sns.FacetGrid(df, col="Region", hue="Producto", height=4, aspect=1)
g.map(sns.scatterplot, "Ventas", "Satisfaccion_Cliente")
g.set_axis_labels("Ventas","Satisfacción del Cliente")
g.set_titles(col_template="Región {col_name}")
g.add_legend()
plt.show();
```

### Respuestas
No se identifica una co-relación entre las variables en ninguna de las regiones.

3- Costos Operativos: ¿Qué regiones tienen los costos operativos más altos? ¿Cómo se comparan estos costos con las ventas y la satisfacción del cliente?

Primero comparemos costos operativos por region

```python
plt.figure(figsize=(10,6))
sns.barplot(data=df, x="Region", y="Costo_Operativo", estimator=np.mean)
plt.title("Costos operativos promedio por región")
plt.xlabel("Región")
plt.ylabel("Costo operativo promedio")
plt.show();
```

La región Este tiene el costo más alto, seguido de la región Sur.

```python
g = sns.FacetGrid(df, col="Region", height=4, aspect=1)
g.map_dataframe(sns.scatterplot, x="Ventas", y="Costo_Operativo", hue="Producto", size="Satisfaccion_Cliente", sizes=(20, 200))
g.set_axis_labels("Ventas","Costo Operativo")
g.set_titles(col_template="Región: {col_name}")
plt.suptitle("Relación entre ventas, costo operativo y satisfacción del cliente por producto", y=1.03, fontsize=16)
g.add_legend()
plt.show();
```

Conclusión: No se evidencian relaciones entre los costos/ventas y la satisfacción del cliente, independientemente del producto
