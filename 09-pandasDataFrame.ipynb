# Pandas - `DataFrame`


Probablemente la estructura de datos **más importante de pandas** sea el __"DataFrame"__. Es una estructura tabular, organiza los datos en filas y columnas, similar a una tabla de base de datos o una hoja de cálculo,  estrechamente integrada con "Series".

## Seleccion Condicional (arrays booleanos)

Funciona igual que en las Series

````python
df
````

Obtener las ciudades que tengan una poblacion mayor a 3 MM.

````python
df['poblacion'] > 3_000_000
````

````python
df.loc[df['poblacion'] > 3_000_000]
````

````python
df.loc[df['poblacion'] > 3_000_000, "latitud"]
````

````python
df.loc[condicion]
````

## Series de Tiempo - Datetime

````python
# Crear un DataFrame con fechas como cadenas de texto (tipo 'object')
data = {'Fecha': ['2024-01-30', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05'],
        'Valor': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)
df
````

````python
df.dtypes
````

````python
# Convertir la columna 'Fecha' de tipo object a datetime
df['Fecha'] = pd.to_datetime(df['Fecha'])
````

````python
# Ver el tipo de la columna 'Fecha' después de la conversión
print("\nDespués de convertir:")
print(df.dtypes)
````

````python
df
````

````python
fecha_comparacion = pd.to_datetime('2024-01-02')
df_mayores = df[df['Fecha'] > fecha_comparacion]
df_mayores
````

````python
df
````

````python
from datetime import datetime
````

````python
fecha_comparacion = datetime(2024, 1, 3)
fecha_comparacion
````

````python
df_mayores = df[df['Fecha'] > fecha_comparacion]
df_mayores
````

````python
df['Fecha_mas_2_dias'] = df['Fecha'] + pd.Timedelta(days=2)
df
````
